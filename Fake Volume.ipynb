```python
import requests
```

```python
res = requests.get('https://kalshi-public-docs.s3.amazonaws.com/reporting/trade_data_2025-09-07.json')
```

```python
df = pd.DataFrame(res.json())
```

```python
df.to_csv('big_day.csv', index=None)
```

```python
df['create_ts'] = pd.to_datetime(df['create_ts'])
```

```python
df['contracts_traded'].sum()
```

```python
ticker_wise = df.groupby('report_ticker')['contracts_traded'].sum()
```

```python
ticker_wise.sort_values()
```
